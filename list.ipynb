```python
l1=list(map(int ,input("enter first list").split()))
l2=list(map(int,input("enter second list").split()))
if len(l1)==len(l2):
 print("both are same length")
else:
 print("different length")
if sum(l1)==sum(l2):
    print("sum are same")
else:
    print("different sum")
if set(l1)&set(l2):
    print("common values:",set(l1)&set(l2))
else:
    print("no common values")
```
